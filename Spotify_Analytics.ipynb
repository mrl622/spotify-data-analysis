## SQL Analysis

```python
from pandasql import sqldf
pql = lambda q:sqldf(q,globals())
```

### Top 10 most popular tracks

```python
q1 = pql(""" 
SELECT track_name,artists,popularity 
FROM df
ORDER BY popularity DESC
LIMIT 10
""")
print(q1)
```

## Average Audio Features BY Genre

```python
q2 = pql ("""
SELECT track_genre,
    ROUND(AVG(popularity),2) AS avg_popularity,
    ROUND(AVG(danceability),2) AS avg_dance,
    ROUND(AVG(energy),2) AS avg_energy
FROM df
GROUP BY track_genre
ORDER BY avg_popularity
LIMIT 10
""")
print(q2)
```

## Explicit vs non-Explicit song

```python
q3 =  pql("""
SELECT explicit,
    COUNT(*) AS total_traks,
    ROUND(AVG(popularity),2) AS avg_popularity
FROM df
GROUP BY explicit	
""")
print(q3)
```

## High Energy + High Popularity Songs

```python
q4 = pql("""
SELECT track_name,artists,popularity,energy	
FROM df
WHERE energy > 0.8 AND popularity > 80 
ORDER BY popularity DESC
LIMIT 10
""")
print(q4)
```
